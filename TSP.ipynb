```python
import numpy as np

def tsp_nearest_neighbor(distances, start_city=0):
    num_cities = distances.shape[0]
    current_city = start_city
    tour = [current_city]
    unvisited_cities = set(range(num_cities))
    unvisited_cities.remove(current_city)

    while unvisited_cities:
        nearest_city = min(unvisited_cities, key=lambda city: distances[current_city, city])
        tour.append(nearest_city)
        unvisited_cities.remove(nearest_city)
        current_city = nearest_city

    tour.append(start_city)

    return tour

city_names = []
for i in range(4):
    city_name = input(f"Enter the name of city {i + 1}: ")
    city_names.append(city_name)

distances = np.zeros((4, 4))

print("Enter the distances between cities (including distances from a city to itself):")
for i in range(4):
    for j in range(4):
        if i != j:
            distance = float(input(f"Distance between {city_names[i]} and {city_names[j]}: "))
            distances[i, j] = distance
        else:
           distance = float(input(f"Distance between {city_names[i]} and {city_names[j]}: "))
           distances[i, j] = distance

start_city_name = input("Enter the starting city: ")
start_city = city_names.index(start_city_name)

tour = tsp_nearest_neighbor(distances, start_city=start_city)

print("Route:", " → ".join(city_names[city] for city in tour))

total_distance = sum(distances[tour[i], tour[i + 1]] for i in range(len(tour) - 1))
print("Total Distance:", total_distance)
```
