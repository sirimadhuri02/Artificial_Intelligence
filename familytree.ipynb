```python
class FamilyTree:
    def __init__(self):
        self.family = {}

    def add_relationship(self, parent, child):

        if parent not in self.family:
            self.family[parent] = {'parents': [], 'children': []}
        if child not in self.family:
            self.family[child] = {'parents': [], 'children': []}

        self.family[parent]['children'].append(child)
        self.family[child]['parents'].append(parent)
        print(f"Relationship added: {parent} -> {child}")

    def query_parents(self, person):

        if person in self.family:
            return self.family[person]['parents']
        return []

    def query_children(self, person):

        if person in self.family:
            return self.family[person]['children']
        return []

    def query_siblings(self, person):

        if person in self.family:
            parents = self.family[person]['parents']
            siblings = []
            for parent in parents:
                siblings.extend(self.family[parent]['children'])
            siblings = list(set(siblings) - {person})
            return siblings
        return []

    def query_ancestors(self, person):
        ancestors = set()
        parents = self.query_parents(person)
        for parent in parents:
            ancestors.add(parent)
            ancestors.update(self.query_ancestors(parent))
        return ancestors

def display_menu():
    print("\nFamily Tree Menu:")
    print("1. Add a parent-child relationship.")
    print("2. Query a person's parents.")
    print("3. Query a person's children.")
    print("4. Query a person's siblings.")
    print("5. Query a person's ancestors.")
    print("6. Exit the program.")

def main():
    family_tree = FamilyTree()

    while True:
        display_menu()
        try:
            choice = int(input("Choose an option: "))
        except ValueError:
            print("Invalid input. Please enter a number between 1 and 6.")
            continue

        if choice == 1:
            parent = input("Enter the parent's name: ")
            child = input("Enter the child's name: ")
            family_tree.add_relationship(parent, child)

        elif choice == 2:
            person = input("Enter the person's name to find their parents: ")
            parents = family_tree.query_parents(person)
            print(f"Parents of {person}: {parents if parents else 'No parents found.'}")

        elif choice == 3:
            person = input("Enter the person's name to find their children: ")
            children = family_tree.query_children(person)
            print(f"Children of {person}: {children if children else 'No children found.'}")

        elif choice == 4:
            person = input("Enter the person's name to find their siblings: ")
            siblings = family_tree.query_siblings(person)
            print(f"Siblings of {person}: {siblings if siblings else 'No siblings found.'}")

        elif choice == 5:
            person = input("Enter the person's name to find their ancestors: ")
            ancestors = family_tree.query_ancestors(person)
            print(f"Ancestors of {person}: {ancestors if ancestors else 'No ancestors found.'}")

        elif choice == 6:
            print("Exiting the program...")
            break

        else:
            print("Invalid choice. Please select a valid option.")

if __name__ == "__main__":
    main()
```
