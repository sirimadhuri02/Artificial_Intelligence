```python
import heapq
def min_max_path(graph, source, target):
    paths = []
    def dfs(node, path, max_weight):
        path.append(node)
        if node == target:
            paths.append((path.copy(), max_weight))
            return

        for neighbor, weight in graph[node].items():
            dfs(neighbor, path, max(max_weight, weight))
        path.pop()
    dfs(source, [],0)






    for i, (path, max_weight) in enumerate(paths):
        print(f"o Path {i + 1}: max({', '.join(map(str, [graph[path[j]][path[j + 1]] for j in range(len(path) - 1)]))})={max_weight}")
    min_max_weight = float('inf')
    best_path = None

    for path,max_weight in paths:
      if max_weight < min_max_weight:
        min_max_weight = max_weight
        best_path = path

    print(f"Select the path with the smallest maximum edge weight:")
    print(f"o The path {'->'.join(best_path)} has the smallest maximum edge weight of {min_max_weight}.")
    return min_max_weight

graph = {}
num_nodes = int(input("Enter the number of nodes in the graph: "))
for _ in range(num_nodes):
    node = input("Enter the node name: ")
    graph[node] = {}
    num_neighbors = int(input(f"Enter the number of neighbors for node {node}: "))
    for _ in range(num_neighbors):
        neighbor = input(f"Enter neighbor for node {node}: ")
        weight = int(input(f"Enter weight of edge between {node} and {neighbor}: "))
        graph[node][neighbor] = weight
source = input("Enter the source node: ")
target = input("Enter the target node: ")
result = min_max_path(graph, source, target)
```
