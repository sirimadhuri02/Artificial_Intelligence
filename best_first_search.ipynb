```python
import heapq

def best_first_search(graph, heuristics, start, end):
    priority_queue = [(heuristics[start], start, [start], 0)]
    visited = set()

    while priority_queue:
        (heuristic, current_node, path, cost) = heapq.heappop(priority_queue)

        if current_node == end:
            return path, cost

        visited.add(current_node)

        if current_node in graph:
            for neighbor, edge_cost in graph[current_node].items():
                if neighbor not in visited:
                    total_cost = cost + edge_cost

                    if neighbor in heuristics:
                        heapq.heappush(priority_queue, (heuristics[neighbor], neighbor, path + [neighbor], total_cost))
                    else:
                        print(f"Warning: Heuristic value not found for neighbor '{neighbor}'. Skipping.")

    return None, None


graph = {}
print("Enter Node Connections and Costs (e.g., A B: 1). Type 'done' when finished:")
while True:
    line = input()
    if line == 'done':
        break
    try:
        parts = line.split(':')
        u, v = parts[0].strip().split(' ', 1)
        cost = int(parts[1].strip())
        graph.setdefault(u, {})[v] = cost
    except (ValueError, IndexError):
        print("Invalid input format. Please use 'Node to Node: Cost'.")

heuristics = {}
print("Enter Heuristic Values (Node Heuristic):")
while True:
    line = input()
    if line == 'done':
        break
    try:
        node, heuristic = line.split()
        heuristics[node] = int(heuristic)
    except (ValueError, IndexError):
        print("Invalid input format. Please use 'Node Heuristic'.")


start_node = input("Enter the start node: ").strip()
end_node = input("Enter the goal node: ").strip()


path, total_cost = best_first_search(graph, heuristics, start_node, end_node)

if path:
    print(f"Found {end_node} with path: {' -> '.join(path)} and total cost: {total_cost}")
else:
    print("Goal not found.")
```
